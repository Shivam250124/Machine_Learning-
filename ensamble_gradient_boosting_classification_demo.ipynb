```python
import pandas as pd
import numpy as np
import random

df = pd.DataFrame(
                  [
                      [6.82, 118, 0],
                      [6.36, 125, 1],
                      [5.39, 99, 1],
                      [5.50, 106, 1],
                      [6.39, 148, 0],
                      [9.13, 148, 1],
                      [7.17, 147, 1],
                      [7.72, 72, 0]
                   ], columns = ['cgpa',	'iq',	'is_placed']
)

df
```

Screenshot 2026-08-17 at 12.41.07 AM.png

```python
# initial prediction

df['pre1(log-odds)'] = np.log(5/3)
df
```

Screenshot 2026-08-17 at 12.42.23 AM.png

```python
# convert log odds to probability
df['pre1(probability)'] = 1/(1+np.exp(-np.log(5/3)))
df
```

```python
# calculating residual for stage 1
df['res1'] = df['is_placed'] - df['pre1(probability)']
df
```

```python
# training the first decision tree
from sklearn.tree import DecisionTreeRegressor

reg1 = DecisionTreeRegressor(max_leaf_nodes=3, random_state=1)

reg1.fit(df.iloc[:,0:2].values, df.iloc[:,-1].values)
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plot_tree(reg1, feature_names=['cgpa','iq'],filled=True, node_ids=True)
plt.show()
```

Here the values in the tree is the difference of the probability(residual).

because we train these trees on residual . So that we have to convert these into log of odds , so theat we can add with previous model .


This code uses the reg1 DecisionTreeRegressor, which was trained earlier, to determine the leaf node each data point in your DataFrame df falls into. Specifically, reg1.apply(df.iloc[:,0:2]) takes the 'cgpa' and 'iq' columns as input for each row and returns the ID of the leaf node that row reaches in the reg1 tree.

```python
df['leaf_entry1'] = reg1.apply(df.iloc[:,0:2])
df
```

Screenshot 2026-08-17 at 12.58.29 AM.png

```python
def return_logs(leaf):
  temp_df = df[df['leaf_entry1'] == leaf]
  num = temp_df['res1'].sum()

  den = sum(temp_df['pre1(probability)'] * (1 - temp_df['pre1(probability)']))
  return round(num/den,2)
```

1st row =  pre2(log-odds) = 0.51 + leaf_entry.(3)

leaf_entry(3)= -0.625-0.625/0.625(1-0.625)+0.625(1-0.625)
=-2.66 # gamma

1st row pre2(log-odds) = 0.51 + (-2.66)= -2.15

```python
df['pre2(log-odds)'] = df['pre1(log-odds)'] + df['leaf_entry1'].apply(return_logs)
df
```

Again convert log of odds into probability

```python
df['pre2(probability)'] = 1/(1+np.exp(-df['pre2(log-odds)']))
df
```

```python
df['res2'] = df['is_placed'] - df['pre2(probability)']
df
```

```python
reg2 = DecisionTreeRegressor(max_leaf_nodes=3, random_state=1)

reg2.fit(df.iloc[:,0:2].values, df.iloc[:,-1].values)
```

```python
plot_tree(reg2, feature_names=['cgpa','iq'],filled=True, node_ids=True)
plt.show()
```

determine the leaf node each data point in your DataFrame

```python
df['leaf_entry2'] = reg2.apply(df.iloc[:,0:2])
df
```

Again calculate gamma

```python
def return_logs(leaf):
  num = df[df['leaf_entry2'] == leaf]['res2'].sum()
  den = sum(df[df['leaf_entry2'] == leaf]['pre2(probability)'] * (1 - df[df['leaf_entry2'] == leaf]['pre2(probability)']))
  return round(num/den,2)
```

Again calculate log of odds

```python
df['pre3(log-odds)'] = df['pre1(log-odds)'] + df['pre2(log-odds)'] + df['leaf_entry2'].apply(return_logs)
```

Convert log odds into probability

```python
df['pre3(probability)'] = 1/(1+np.exp(-df['pre3(log-odds)']))
df
```

```python
df['res_final'] = df['is_placed'] - df['pre3(probability)']

df[['res1','res2','res_final']]
```
